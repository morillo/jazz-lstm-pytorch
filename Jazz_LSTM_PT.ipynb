---

### TensorFlow vs PyTorch — the key differences

| | TensorFlow / Keras | PyTorch |
|---|---|---|
| Model definition | Functional API | `nn.Module` subclass |
| LSTM cell | `LSTM(return_state=True)` | `nn.LSTMCell` |
| Dense layer | `Dense(n, activation='softmax')` | `nn.Linear` + manual softmax |
| Training | `model.fit()` | manual loop with `loss.backward()` |
| Loss | `categorical_crossentropy` | `F.nll_loss` over log-probabilities |

The math is identical. The framework exposes it differently.
PyTorch forces you to be explicit — which is exactly what makes it
a favourite for research and for understanding what is really happening inside the model.

---

### Why this matters beyond music

The autoregressive loop you will see here — predict one token, feed it back,
predict the next — is the **heartbeat of every voice AI agent** in production.
In a voice system the tokens are acoustic units of speech rather than musical notes.
The loop is the same. Understanding it here means understanding it everywhere.

## Part 1 — The Dataset

### Tokenizing music

Music is continuous. Neural networks are discrete. To bridge this gap we convert
the jazz corpus into a sequence of **musical values** — each value encodes a note,
its duration, and its harmonic context as a single token in a vocabulary of 90 symbols.

This is directly analogous to how voice AI systems tokenize speech: raw audio
becomes a stream of discrete acoustic units, and a sequence model learns to predict
the next unit given all previous ones. The vocabulary is different; the architecture is not.

### Converting to PyTorch tensors

After loading the corpus through the same music21 pipeline used in the TensorFlow version,
we convert the NumPy arrays to PyTorch tensors and move them to the available compute device.


````python
X_np, Y_np, n_values, indices_values, chords = load_music_utils('./data/original_metheny.mid')

print(f'Training examples (m) : {X_np.shape[0]}')
print(f'Sequence length  (Tx) : {X_np.shape[1]}')
print(f'Unique values         : {n_values}')
print(f'Shape of X            : {X_np.shape}')

X = torch.tensor(X_np, dtype=torch.float32).to(device)
Y = [torch.tensor(np.array(y), dtype=torch.float32).to(device) for y in Y_np]
Tx, m, n_a = X.shape[1], X.shape[0], 64

print(f'Tensors on : {X.device}  |  dtype: {X.dtype}')
````

## Part 2 — The Architecture

### Building the model in PyTorch

In TensorFlow we used the Keras Functional API to wire up layers declaratively.
In PyTorch we subclass `nn.Module` and write the forward pass ourselves.
This makes the computation explicit and easy to inspect.

The architecture is identical to the TensorFlow version:

| Step | TensorFlow | PyTorch |
|---|---|---|
| Define LSTM | `LSTM(n_a, return_state=True)` | `nn.LSTMCell(n_values, n_a)` |
| Define projection | `Dense(n_values, softmax)` | `nn.Linear(n_a, n_values)` |
| Loop over time | `for t in range(Tx)` | `for t in range(X.shape[1])` |
| Update state | `_, a, c = LSTM_cell(x, [a, c])` | `a, c = lstm_cell(x, (a, c))` |
| Predict | `densor(a)` | `softmax(linear(a))` |

The `DJModel` class contains both the **training forward pass** and a separate
**`generate` method** for autoregressive inference — keeping training and generation
logic clearly separated.


````python
class DJModel(nn.Module):
    """
    Jazz LSTM: shared-weight LSTM unrolled over Tx time steps.

    The same lstm_cell and linear weights are used at every time step.
    This is the PyTorch equivalent of the shared-weight pattern in the
    TensorFlow version (global LSTM_cell + densor layers).

    Attributes
    ----------
    lstm_cell : nn.LSTMCell(n_values, n_a)   <- Keras: LSTM(n_a, return_state=True)
    linear    : nn.Linear(n_a, n_values)      <- Keras: Dense(n_values)
    """

    def __init__(self, n_values, n_a):
        super().__init__()
        self.n_a       = n_a
        self.n_values  = n_values
        self.lstm_cell = nn.LSTMCell(n_values, n_a)
        self.linear    = nn.Linear(n_a, n_values)

    def forward(self, X, a0, c0):
        """
        Training forward pass.
        X  : (m, Tx, n_values)  — batch of musical sequences
        a0 : (m, n_a)           — initial hidden state (zeros)
        c0 : (m, n_a)           — initial cell state (zeros)
        Returns list of Tx tensors, each (m, n_values).
        """
        a, c, outputs = a0, c0, []
        for t in range(X.shape[1]):
            a, c = self.lstm_cell(X[:, t, :], (a, c))
            outputs.append(torch.softmax(self.linear(a), dim=-1))
        return outputs

    def generate(self, x0, a0, c0, Ty=50):
        """
        Autoregressive inference.
        Each predicted value becomes the input for the next step.
        This is the same loop as music_inference_model in the TensorFlow version.

        x0 : (1, n_values)  — zero-vector seed (silence before the first note)
        Returns list of Ty tensors, each (1, n_values).
        """
        self.eval()
        with torch.no_grad():
            a, c, x, outputs = a0, c0, x0, []
            for _ in range(Ty):
                a, c = self.lstm_cell(x, (a, c))
                out  = torch.softmax(self.linear(a), dim=-1)
                outputs.append(out)
                x = F.one_hot(torch.argmax(out, dim=-1), num_classes=self.n_values).float()
        return outputs


model = DJModel(n_values=n_values, n_a=n_a).to(device)
print(model)
print(f'Parameters : {sum(p.numel() for p in model.parameters()):,}')
````

---

## Part 3 — Training

### The manual training loop

This is where PyTorch differs most visibly from Keras.
There is no `model.fit()`. We write the loop ourselves:

```python
for epoch in range(100):
    predictions = model(X, a0, c0)        # forward pass
    loss = sum(nll_loss(pred, target)      # loss over all Tx steps
               for pred, target in zip(predictions, Y))
    loss.backward()                        # compute gradients
    optimizer.step()                       # update weights
```

This explicitness is a feature, not a bug. Every operation is visible.
Every gradient flows through code you can read and modify.

Watch the loss fall. The model is absorbing jazz, one epoch at a time.


````python
EPOCHS, LR = 100, 0.01
optimizer = Adam(model.parameters(), lr=LR, betas=(0.9, 0.999), weight_decay=0.01)
losses = []
model.train()

print(f'Epochs        : {EPOCHS}')
print(f'Learning rate : {LR}')
print(f'Parameters    : {sum(p.numel() for p in model.parameters()):,}')
print('-' * 40)

for epoch in range(EPOCHS):
    a0 = torch.zeros(m, n_a, device=device)
    c0 = torch.zeros(m, n_a, device=device)
    optimizer.zero_grad()
    preds = model(X, a0, c0)
    loss  = torch.tensor(0.0, device=device)
    for t in range(Tx):
        loss = loss + F.nll_loss(torch.log(preds[t] + 1e-8), torch.argmax(Y[t], dim=-1))
    loss.backward()
    optimizer.step()
    losses.append(loss.item())
    if (epoch + 1) % 10 == 0 or epoch == 0:
        print(f'Epoch {epoch+1:>3}/{EPOCHS}  loss: {loss.item():.4f}')

print('-' * 40)
print(f'Start: {losses[0]:.4f}  |  End: {losses[-1]:.4f}')
````

````python
plt.figure(figsize=(8, 4))
plt.plot(losses, color='darkorange', linewidth=2)
plt.title('Training Loss — Jazz LSTM (PyTorch)')
plt.xlabel('Epoch')
plt.ylabel('NLL Loss')
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
````

## Part 4 — Generating Music

### The moment of creation

Training is over. We call `model.generate()` with a zero-vector seed —
silence, the blank canvas before the first note.

The model runs its autoregressive loop: predict a note, play it, feed it back,
predict the next. Fifty steps. Fifty decisions. A solo that has never existed before.

The output tensors live on the compute device. We move them back to CPU
before passing them to the music21 post-processing pipeline, which decodes
the indices back into actual notes and writes a MIDI file.


````python
# Seed the model with silence and generate 50 notes
x0 = torch.zeros(1, n_values, device=device)
a0 = torch.zeros(1, n_a, device=device)
c0 = torch.zeros(1, n_a, device=device)

raw = model.generate(x0, a0, c0, Ty=50)

# Move to CPU for the music21 pipeline
pred_np = [o.cpu().numpy() for o in raw]
indices = np.array([np.argmax(o) for o in pred_np])

print(f'Generated {len(indices)} note indices')
print(f'First 10  : {indices[:10].tolist()}')
print(f'Unique values used: {len(set(indices.tolist()))} / {n_values}')
````

````python
# Convert to one-hot for the music21 pipeline
results = np.eye(n_values)[indices]
print(f'results shape: {results.shape}')
````

### Post-processing and MIDI rendering

The `PTInferenceAdapter` bridges the PyTorch model into the same music21 pipeline
used by the TensorFlow version. This is a clean separation of concerns:
the neural network handles generation, the post-processor handles musicality.

The post-processor applies jazz grammar rules — pruning repeated notes,
smoothing dissonant intervals, anchoring the melody to the underlying chord changes.
This is analogous to the prosody normalization layer in a voice AI system:
the model generates tokens, a downstream processor shapes them into something
that sounds natural and human.


````python
class PTInferenceAdapter:
    """
    Wraps the PyTorch model to match the interface expected by
    generate_music() in data_utils.py (which was written for TensorFlow).

    By implementing .predict() and exposing .n_values, this adapter
    lets the same music21 pipeline work with both frameworks without
    any changes to the post-processing code.
    """
    def __init__(self, model, device, n_values, n_a, Ty=50):
        self.pt_model = model
        self.device   = device
        self.n_values = n_values
        self.n_a      = n_a
        self.Ty       = Ty

    def predict(self, inputs):
        x0, a0, c0 = inputs
        x0 = torch.tensor(x0, dtype=torch.float32).squeeze(1).to(self.device)
        a0 = torch.tensor(a0, dtype=torch.float32).to(self.device)
        c0 = torch.tensor(c0, dtype=torch.float32).to(self.device)
        return [o.cpu().numpy() for o in self.pt_model.generate(x0, a0, c0, Ty=self.Ty)]


adapter    = PTInferenceAdapter(model, device, n_values, n_a, Ty=50)
out_stream = generate_music(adapter, indices_values, chords)
print('MIDI saved to ./output/my_music.midi')
````
